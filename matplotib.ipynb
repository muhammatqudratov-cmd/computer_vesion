```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([20, 25, 30, 45])

plt.plot(x, y)
plt.title("Basic Plot Line")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
x  = np.array([1, 2, 3, 4])
y = np.array([7, 10, 12, 9])

plt.plot(y, linestyle='dotted', color='green', marker='o')
plt.title("Styled Line Exwmple")
plt.show()
```

```python
x  = np.array([1, 2, 3, 4])
y1 = np.array([7, 10, 12, 9])
y2 = np.array([5, 8, 6, 11])

plt.plot(x, y1, label='Line 1', color='green', marker='s')
plt.plot(x, y2, label='Line 2', color='blue', linestyle='--', marker='o')
plt.title("Multiple Lines Exemple")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.grid(True)
plt.legend()
plt.show()
```

# **Bar Chart with Numpy and Matplotib**

Scatter Plot with Numpy and **Matplotib**

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 6, 3, 7])

plt. scatter(x , y, color='purple')
plt.title('Scatter Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

# Histogram with Nimpy and Matpltib

```python
data = np.array([3, 4, 5, 6, 8, 9, 10])

plt.hist(data, bins=5, color='darkblue', edgecolor="white")
plt.title("Histogram Exemple")
plt.xlabel('Values')
plt.ylabel('Count')
plt.show()
```

Subplots(Multiple Graphs Together)

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])

plt.figure(figsize=(8, 4))

# Plot 1
plt.subplot(1, 2, 1)
plt.plot(x, y1, color='red')
plt.title('Plot 1')

# Plot 2
plt.subplot(1, 2, 2)
plt.bar(x, y2, color='blue')
plt.title('Plot 2')

plt.tight_layout()
plt.show()
```
